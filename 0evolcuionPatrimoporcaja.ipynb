```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# ==============================
# PARÁMETROS
# ==============================

tipo_cambio = 7300  # Gs/USD
years = list(range(2011, 2026))
years_completos = list(range(2011, 2027))  # Incluye 2026

# ==============================
# DATOS DE INGRESOS Y GASTOS (millones Gs)
# ==============================

ingresos = {
    "Administración Pública": [558028,833889,875256,981794,927346,806454,843915,946621,1003088,1021055,1005387,1492746,1603068,1175950,1217113],
    "Magistrados Judiciales": [62100,68363,74309,82306,74913,46387,47639,48596,52492,67204,69292,116076,131326,93952,104650],
    "Magisterio Nacional": [424384,517165,540711,556635,595651,551489,610754,734177,814419,798002,847189,957696,1176179,1133104,1217098],
    "Docentes Universitarios": [56563,76935,86116,91000,91510,94854,96174,100632,114674,123651,126160,152017,161183,145808,152360],
    "Fuerzas Armadas": [75704,105863,140786,122487,114102,104922,118688,126948,134210,144559,153814,192294,209040,206171,222958],
    "Fuerzas Policiales": [105310,185203,188900,194102,198209,185211,210329,221542,235841,253414,263018,311007,340251,344604,399713]
}

gastos = {
    "Administración Pública": [254815,267744,280166,310374,349071,377159,428184,498116,549247,554717,598718,750090,870743,990179,1102420],
    "Magistrados Judiciales": [15843,16494,19457,19240,22509,22166,25545,36107,46412,48704,60983,90347,108080,134405,146442],
    "Magisterio Nacional": [505043,553423,600485,652908,727710,787892,886801,1027018,1198139,1410033,1483533,1657622,1913773,2104391,2390176],
    "Docentes Universitarios": [33399,36495,39554,44820,49378,57922,72444,83509,93998,101044,106875,137322,164802,181923,197102],
    "Fuerzas Armadas": [274230,363833,418033,436157,490619,510850,550049,607036,636130,652164,655760,706862,787957,859713,910499],
    "Fuerzas Policiales": [190106,290594,352970,379254,432605,462387,506337,571657,616791,658870,691869,760326,915686,1048102,1127156]
}

# ==============================
# DATOS DE PATRIMONIO NETO 2023 POR SECTOR (ÚNICO AÑO CONOCIDO)
# ==============================
"""
SEGÚN PÁGINAS 78-80 DEL INFORME:
- Administración Pública: USD 682.83 MM (Pág 78)
- Magistrados Judiciales: USD 51.04 MM (Pág 79)  
- Magisterio Nacional: USD 7.02 MM (Pág 78)
- Docentes Universitarios: USD 0.17 MM (Pág 79)
- Fuerzas Armadas: USD 0.00 MM (Pág 80)
- Fuerzas Policiales: USD 0.00 MM (Pág 80)

TOTAL PROGRAMA CIVIL: USD 741 MM (Pág 75)
"""

patrimonio_neto_2023_usd_sector = {
    "Administración Pública": 682.83,
    "Magistrados Judiciales": 51.04,
    "Magisterio Nacional": 7.02,
    "Docentes Universitarios": 0.17,
    "Fuerzas Armadas": 0.00,
    "Fuerzas Policiales": 0.00
}

# Patrimonio total 2023 (suma de todos los sectores)
patrimonio_total_2023_usd = sum(patrimonio_neto_2023_usd_sector.values())
patrimonio_total_2023_gs = patrimonio_total_2023_usd * tipo_cambio

# ==============================
# CREAR DATAFRAMES DE INGRESOS, GASTOS Y RESULTADOS
# ==============================

df_ing = pd.DataFrame(ingresos, index=years)
df_gas = pd.DataFrame(gastos, index=years)

# Agregar totales
df_ing["TOTAL"] = df_ing.sum(axis=1)
df_gas["TOTAL"] = df_gas.sum(axis=1)

# Calcular resultados por sector
resultado_sector = df_ing - df_gas
resultado_sector["TOTAL"] = resultado_sector.sum(axis=1)

# ==============================
# PROYECTAR 2026 (SOLO PARA TOTAL)
# ==============================

def proyectar_lineal(df, columna, años=[2024, 2025]):
    """Proyecta usando tendencia lineal"""
    x = años
    y = [df.loc[a, columna] for a in años]
    z = np.polyfit(x, y, 1)
    p = np.poly1d(z)
    return p(2026)

# Proyectar ingresos y gastos totales 2026
df_ing.loc[2026, "TOTAL"] = proyectar_lineal(df_ing, "TOTAL")
df_gas.loc[2026, "TOTAL"] = proyectar_lineal(df_gas, "TOTAL")
resultado_sector.loc[2026, "TOTAL"] = df_ing.loc[2026, "TOTAL"] - df_gas.loc[2026, "TOTAL"]

# ==============================
# RECONSTRUCCIÓN DEL PATRIMONIO TOTAL (SÍ ES POSIBLE)
# ==============================
"""
SÍ podemos reconstruir el patrimonio TOTAL porque:
- Tenemos el patrimonio TOTAL 2023 (USD 741 MM)
- Tenemos el resultado TOTAL anual (ingresos - gastos)
- El patrimonio TOTAL sigue la ecuación: P(t) = P(t-1) + R(t)

NO podemos reconstruir patrimonio por sector porque:
- Desconocemos las transferencias entre regímenes
- Los resultados sectoriales NO se acumulan independientemente
- El fondo es común, no hay cuentas separadas
"""

patrimonio_total = {}
patrimonio_total[2023] = patrimonio_total_2023_gs

print("\n" + "="*80)
print("✅ RECONSTRUCCIÓN DEL PATRIMONIO TOTAL (2011-2026)")
print("="*80)
print(f"Patrimonio TOTAL 2023: Gs {patrimonio_total[2023]/1000:,.0f} mil MM | USD {patrimonio_total[2023]/tipo_cambio:,.0f} MM (BASE INFORME)")

# Reconstruir hacia atrás (2011-2022)
for y in range(2022, 2010, -1):
    patrimonio_total[y] = patrimonio_total[y+1] - resultado_sector.loc[y+1, "TOTAL"]
    print(f"Patrimonio TOTAL {y}: Gs {patrimonio_total[y]/1000:,.0f} mil MM | "
          f"USD {patrimonio_total[y]/tipo_cambio:,.0f} MM | "
          f"Resultado {y+1}: Gs {resultado_sector.loc[y+1, 'TOTAL']/1000:,.0f} mil MM")

# Proyectar hacia adelante (2024-2026)
for y in [2024, 2025, 2026]:
    patrimonio_total[y] = patrimonio_total[y-1] + resultado_sector.loc[y, "TOTAL"]
    print(f"Patrimonio TOTAL {y}: Gs {patrimonio_total[y]/1000:,.0f} mil MM | "
          f"USD {patrimonio_total[y]/tipo_cambio:,.0f} MM | "
          f"Resultado {y}: Gs {resultado_sector.loc[y, 'TOTAL']/1000:,.0f} mil MM")

# ==============================
# DATAFRAMES PARA VISUALIZACIÓN
# ==============================

# DataFrame de patrimonio total
df_pat_total = pd.DataFrame({
    'Año': list(patrimonio_total.keys()),
    'Patrimonio_MMGs': list(patrimonio_total.values()),
    'Patrimonio_MMUSD': [v/tipo_cambio for v in patrimonio_total.values()]
}).set_index('Año').sort_index()

# DataFrame de ingresos y gastos totales (completo)
df_ing_total = df_ing[["TOTAL"]].copy()
df_gas_total = df_gas[["TOTAL"]].copy()

# ==============================
# CONFIGURACIÓN DE ESTILOS
# ==============================

plt.style.use('seaborn-v0_8-darkgrid')
plt.rcParams['font.size'] = 11
plt.rcParams['axes.labelsize'] = 12
plt.rcParams['axes.titlesize'] = 14
plt.rcParams['xtick.labelsize'] = 10
plt.rcParams['ytick.labelsize'] = 10
plt.rcParams['legend.fontsize'] = 10
plt.rcParams['figure.titlesize'] = 16

colores_sectores = {
    "Administración Pública": "#2E86AB",
    "Magistrados Judiciales": "#F39C12",
    "Magisterio Nacional": "#229954",
    "Docentes Universitarios": "#E67E22",
    "Fuerzas Armadas": "#5D6D7E",
    "Fuerzas Policiales": "#884EA0"
}

# ==============================
# GRÁFICO 1: PATRIMONIO TOTAL (EVOLUCIÓN COMPLETA)
# ==============================

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Patrimonio total en Gs
ax1.plot(df_pat_total.index, df_pat_total["Patrimonio_MMGs"]/1000, 'o-', 
         linewidth=3, markersize=8, color='#1F618D', label='Patrimonio Total')
ax1.fill_between(df_pat_total.index, df_pat_total["Patrimonio_MMGs"]/1000, alpha=0.2, color='#1F618D')
ax1.axvline(x=2023, color='red', linestyle='--', linewidth=2, alpha=0.7, label='Año base 2023 (informe)')
ax1.axhline(y=0, color='black', linestyle='-', linewidth=0.5, alpha=0.5)
ax1.set_title('PATRIMONIO NETO TOTAL\n(miles de millones de Gs)', fontsize=14, fontweight='bold')
ax1.set_xlabel('Año', fontsize=12, fontweight='bold')
ax1.set_ylabel('Miles de Millones de Gs', fontsize=12, fontweight='bold')
ax1.legend(loc='best', fontsize=11)
ax1.grid(True, alpha=0.3)
ax1.set_xlim(2010, 2027)

# Patrimonio total en USD
ax2.plot(df_pat_total.index, df_pat_total["Patrimonio_MMUSD"], 's-', 
         linewidth=3, markersize=8, color='#117A65', label='Patrimonio Total')
ax2.fill_between(df_pat_total.index, df_pat_total["Patrimonio_MMUSD"], alpha=0.2, color='#117A65')
ax2.axvline(x=2023, color='red', linestyle='--', linewidth=2, alpha=0.7, label='Año base 2023 (informe)')
ax2.axhline(y=0, color='black', linestyle='-', linewidth=0.5, alpha=0.5)
ax2.set_title('PATRIMONIO NETO TOTAL\n(millones de USD)', fontsize=14, fontweight='bold')
ax2.set_xlabel('Año', fontsize=12, fontweight='bold')
ax2.set_ylabel('Millones de USD', fontsize=12, fontweight='bold')
ax2.legend(loc='best', fontsize=11)
ax2.grid(True, alpha=0.3)
ax2.set_xlim(2010, 2027)

# Anotación valor 2023
ax2.annotate(f'USD {patrimonio_total[2023]/tipo_cambio:,.0f} MM',
             xy=(2023, patrimonio_total[2023]/tipo_cambio),
             xytext=(2018, patrimonio_total[2023]/tipo_cambio * 1.5),
             arrowprops=dict(arrowstyle='->', color='darkred', lw=1.5),
             fontsize=11, fontweight='bold',
             bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.8))

plt.suptitle('CAJA FISCAL - EVOLUCIÓN DEL PATRIMONIO NETO TOTAL (RECONSTRUCCIÓN VÁLIDA)', 
             fontsize=16, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()

# ==============================
# GRÁFICO 2: PATRIMONIO POR SECTOR (SOLO 2023 - DATO CONOCIDO)
# ==============================

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 7))

# Preparar datos
sectores = list(patrimonio_neto_2023_usd_sector.keys())
valores = list(patrimonio_neto_2023_usd_sector.values())
colores = [colores_sectores[s] for s in sectores]

# Filtrar sectores con patrimonio > 0 para el gráfico de torta
sectores_positivos = [s for s, v in zip(sectores, valores) if v > 0]
valores_positivos = [v for v in valores if v > 0]
colores_positivos = [colores_sectores[s] for s in sectores_positivos]

# Gráfico de torta - Solo sectores CON patrimonio
if valores_positivos:
    wedges, texts, autotexts = ax1.pie(valores_positivos, 
                                       labels=sectores_positivos, 
                                       autopct=lambda pct: f'USD {pct*sum(valores_positivos)/100:,.0f} MM\n({pct:.1f}%)',
                                       colors=colores_positivos, 
                                       startangle=90, 
                                       explode=[0.05] * len(sectores_positivos),
                                       textprops={'fontsize': 11, 'fontweight': 'bold'})
    
    ax1.set_title('DISTRIBUCIÓN DEL PATRIMONIO NETO POR SECTOR\nAÑO 2023 (ÚNICO AÑO CONOCIDO)', 
                  fontsize=14, fontweight='bold')

# Gráfico de barras - Todos los sectores
bars = ax2.barh(sectores, valores, color=colores, edgecolor='black', linewidth=0.5)
ax2.axvline(x=0, color='black', linestyle='-', linewidth=0.5)
ax2.set_title('PATRIMONIO NETO POR SECTOR - AÑO 2023\n(DATO DEL INFORME, NO RECONSTRUIBLE PARA OTROS AÑOS)', 
              fontsize=14, fontweight='bold')
ax2.set_xlabel('Millones de USD', fontsize=12, fontweight='bold')

# Agregar valores a las barras
for bar, valor in zip(bars, valores):
    width = bar.get_width()
    ax2.text(width + 5, bar.get_y() + bar.get_height()/2, 
            f'USD {valor:,.0f} MM' if valor > 0 else 'USD 0',
            ha='left', va='center', fontweight='bold', fontsize=10)

plt.suptitle('PATRIMONIO NETO (RECURSOS EXCEDENTES) - CAJA FISCAL 2023\n'
             '⚠️ ESTOS SON LOS ÚNICOS DATOS CONFIABLES POR SECTOR - NO RECONSTRUIR HACIA ATRÁS ⚠️',
             fontsize=14, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()

# ==============================
# GRÁFICO 3: INGRESOS Y GASTOS POR SECTOR (EVOLUCIÓN SÍ VÁLIDA)
# ==============================
"""
ESTOS SÍ se pueden graficar porque son datos observados, no reconstruidos
"""

fig, axes = plt.subplots(2, 3, figsize=(20, 12))
axes = axes.flatten()

for idx, sector in enumerate(ingresos.keys()):
    ax = axes[idx]
    
    ing = df_ing[sector] / 1000
    gas = df_gas[sector] / 1000
    
    ax.plot(ing.index, ing, 'o-', linewidth=2.5, markersize=5, 
            color='green', label='Ingresos')
    ax.plot(gas.index, gas, 's-', linewidth=2.5, markersize=5, 
            color='red', label='Gastos')
    
    ax.fill_between(ing.index, ing, gas,
                    where=ing > gas, interpolate=True, alpha=0.2, color='green')
    ax.fill_between(ing.index, ing, gas,
                    where=ing <= gas, interpolate=True, alpha=0.2, color='red')
    
    ax.set_title(f'{sector}', fontweight='bold', fontsize=13)
    ax.set_xlabel('Año', fontsize=11)
    ax.set_ylabel('Miles de Millones de Gs', fontsize=11)
    ax.legend(loc='best', fontsize=9)
    ax.grid(True, alpha=0.3)
    ax.set_xlim(2010, 2027)
    ax.axvline(x=2023, color='gray', linestyle='--', linewidth=1, alpha=0.5)

plt.suptitle('EVOLUCIÓN DE INGRESOS VS GASTOS POR SECTOR (2011-2026)\n'
             '✅ DATOS OBSERVADOS (2011-2025) Y PROYECCIÓN 2026\n'
             'Valores en miles de millones de guaraníes', 
             fontsize=18, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ==============================
# GRÁFICO 4: INGRESOS Y GASTOS TOTALES
# ==============================

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Ingresos totales
ax1.plot(df_ing.index, df_ing["TOTAL"]/1000, 'o-', 
         linewidth=3, markersize=8, color='#229954', label='Ingresos Totales')
ax1.fill_between(df_ing.index, df_ing["TOTAL"]/1000, alpha=0.2, color='#229954')
ax1.set_title('EVOLUCIÓN DE INGRESOS TOTALES', fontsize=14, fontweight='bold')
ax1.set_xlabel('Año', fontsize=12, fontweight='bold')
ax1.set_ylabel('Miles de Millones de Gs', fontsize=12, fontweight='bold')
ax1.grid(True, alpha=0.3)
ax1.set_xlim(2010, 2027)
ax1.axvline(x=2023, color='red', linestyle='--', linewidth=1, alpha=0.7, label='2023')

# Gastos totales
ax2.plot(df_gas.index, df_gas["TOTAL"]/1000, 's-', 
         linewidth=3, markersize=8, color='#C0392B', label='Gastos Totales')
ax2.fill_between(df_gas.index, df_gas["TOTAL"]/1000, alpha=0.2, color='#C0392B')
ax2.set_title('EVOLUCIÓN DE GASTOS TOTALES', fontsize=14, fontweight='bold')
ax2.set_xlabel('Año', fontsize=12, fontweight='bold')
ax2.set_ylabel('Miles de Millones de Gs', fontsize=12, fontweight='bold')
ax2.grid(True, alpha=0.3)
ax2.set_xlim(2010, 2027)
ax2.axvline(x=2023, color='red', linestyle='--', linewidth=1, alpha=0.7, label='2023')

plt.suptitle('CAJA FISCAL - EVOLUCIÓN DE INGRESOS Y GASTOS TOTALES', 
             fontsize=16, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()

# ==============================
# GRÁFICO 5: COMPOSICIÓN DE INGRESOS Y GASTOS (STACKED AREA)
# ==============================

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(18, 7))

# Stacked area de ingresos
ingresos_stack = df_ing[list(ingresos.keys())] / 1000
ax1.stackplot(ingresos_stack.index, ingresos_stack.T, 
              labels=ingresos_stack.columns, 
              colors=[colores_sectores[s] for s in ingresos_stack.columns],
              alpha=0.8)
ax1.set_title('COMPOSICIÓN DE INGRESOS POR SECTOR', fontsize=14, fontweight='bold')
ax1.set_xlabel('Año', fontsize=12, fontweight='bold')
ax1.set_ylabel('Miles de Millones de Gs', fontsize=12, fontweight='bold')
ax1.legend(loc='upper left', bbox_to_anchor=(1, 1), fontsize=10)
ax1.grid(True, alpha=0.3)
ax1.set_xlim(2010, 2027)

# Stacked area de gastos
gastos_stack = df_gas[list(gastos.keys())] / 1000
ax2.stackplot(gastos_stack.index, gastos_stack.T, 
              labels=gastos_stack.columns, 
              colors=[colores_sectores[s] for s in gastos_stack.columns],
              alpha=0.8)
ax2.set_title('COMPOSICIÓN DE GASTOS POR SECTOR', fontsize=14, fontweight='bold')
ax2.set_xlabel('Año', fontsize=12, fontweight='bold')
ax2.set_ylabel('Miles de Millones de Gs', fontsize=12, fontweight='bold')
ax2.legend(loc='upper left', bbox_to_anchor=(1, 1), fontsize=10)
ax2.grid(True, alpha=0.3)
ax2.set_xlim(2010, 2027)

plt.suptitle('CAJA FISCAL - COMPOSICIÓN DE INGRESOS Y GASTOS POR SECTOR\n'
             '✅ DATOS OBSERVADOS (2011-2025) Y PROYECCIÓN 2026', 
             fontsize=16, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()

# ==============================
# GRÁFICO 6: SUPERÁVIT/DÉFICIT POR SECTOR
# ==============================

fig, axes = plt.subplots(2, 3, figsize=(20, 12))
axes = axes.flatten()

for idx, sector in enumerate(resultado_sector.columns):
    if sector == "TOTAL":
        continue
    ax = axes[idx]
    
    res = resultado_sector[sector] / 1000
    
    colors_bar = ['green' if x > 0 else 'red' for x in res]
    bars = ax.bar(res.index, res, color=colors_bar, edgecolor='black', linewidth=0.5, alpha=0.7)
    
    ax.axhline(y=0, color='black', linestyle='-', linewidth=0.5)
    ax.set_title(f'{sector}', fontweight='bold', fontsize=13)
    ax.set_xlabel('Año', fontsize=11)
    ax.set_ylabel('Miles de Millones de Gs', fontsize=11)
    ax.grid(True, alpha=0.3, axis='y')
    ax.set_xlim(2010, 2027)
    ax.axvline(x=2023, color='gray', linestyle='--', linewidth=1, alpha=0.5)

plt.suptitle('SUPERÁVIT/DÉFICIT ANUAL POR SECTOR (2011-2026)\n'
             '✅ DATOS OBSERVADOS (2011-2025) Y PROYECCIÓN 2026', 
             fontsize=18, fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ==============================
# GRÁFICO 7: SUPERÁVIT/DÉFICIT TOTAL
# ==============================

fig, ax = plt.subplots(figsize=(14, 7))

res_total = resultado_sector["TOTAL"] / 1000
colors_bar = ['green' if x > 0 else 'red' for x in res_total]
bars = ax.bar(res_total.index, res_total, color=colors_bar, edgecolor='black', linewidth=0.5, alpha=0.8)

ax.axhline(y=0, color='black', linestyle='-', linewidth=0.5)
ax.set_title('SUPERÁVIT/DÉFICIT TOTAL DE LA CAJA FISCAL (2011-2026)', 
             fontsize=16, fontweight='bold')
ax.set_xlabel('Año', fontsize=13, fontweight='bold')
ax.set_ylabel('Miles de Millones de Gs', fontsize=13, fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
ax.set_xlim(2010, 2027)
ax.axvline(x=2023, color='red', linestyle='--', linewidth=1.5, alpha=0.7, label='2023')

# Agregar valores en las barras
for bar in bars:
    height = bar.get_height()
    if abs(height) > 100:
        ax.text(bar.get_x() + bar.get_width()/2, 
               height + (20 if height > 0 else -60),
               f'{height:,.0f}K', 
               ha='center', va='bottom' if height > 0 else 'top',
               fontsize=9, fontweight='bold', rotation=45)

plt.tight_layout()
plt.show()

# ==============================
# RESUMEN FINAL - QUÉ SE PUEDE Y QUÉ NO SE PUEDE HACER
# ==============================

print("\n" + "="*80)
print("📊 RESUMEN EJECUTIVO - LIMITACIONES DE LA RECONSTRUCCIÓN")
print("="*80)

print(f"""
✅ LO QUE SÍ PODEMOS HACER (DATOS CONFIABLES):
───────────────────────────────────────────────────
1. Patrimonio TOTAL 2011-2026 → RECONSTRUCCIÓN VÁLIDA
   • Tenemos patrimonio TOTAL 2023 (USD 741 MM)
   • Tenemos resultado TOTAL anual
   • Ecuación: P(t) = P(t-1) + R(t) funciona para el TOTAL

2. Ingresos por sector 2011-2025 → DATOS OBSERVADOS
3. Gastos por sector 2011-2025 → DATOS OBSERVADOS  
4. Resultado por sector 2011-2025 → DATOS OBSERVADOS
5. Proyección 2026 → ESTIMACIÓN (con tendencia)

❌ LO QUE NO PODEMOS HACER (SIN DATOS DE TRANSFERENCIAS):
───────────────────────────────────────────────────
1. Patrimonio por sector ANTES DE 2023 → IMPOSIBLE
   • Desconocemos cómo se distribuyeron los superávits
   • No hay registros de transferencias entre regímenes
   • El fondo es común, no hay cuentas separadas

2. Patrimonio por sector DESPUÉS DE 2023 → SOLO CON SUPUESTOS ARBITRARIOS

📌 CONCLUSIÓN:
───────────────────────────────────────────────────
• Gráfico de patrimonio TOTAL → VÁLIDO para 2011-2026
• Gráfico de patrimonio por sector → SOLO para 2023 (dato del informe)
• Gráficos de ingresos/gastos/resultado por sector → VÁLIDOS para 2011-2026

PATRIMONIO TOTAL RECONSTRUIDO:
   ├─ 2011: USD {patrimonio_total[2011]/tipo_cambio:,.0f} MM
   ├─ 2023: USD {patrimonio_total[2023]/tipo_cambio:,.0f} MM (BASE INFORME)
   └─ 2026: USD {patrimonio_total[2026]/tipo_cambio:,.0f} MM (PROYECTADO)
""")
print("="*80)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression

# ==============================
# PARÁMETROS
# ==============================
tipo_cambio = 7300          # Gs/USD
tasa_interes = 0.09          # 9% anual
tasa_base = 0.16             # tasa de cotización original (16%)
tasas_reforma = [0.19, 0.22, 0.25]  # tasas a simular desde 2026
años_hist = list(range(2011, 2026))   # 2011-2025
años_proy = list(range(2026, 2031))   # 2026-2030
años_total = list(range(2011, 2031))

# ==============================
# DATOS REALES DE INGRESOS Y GASTOS TOTALES (2011-2025)
# ==============================
ingresos_reales = [
    1_282_089, 1_787_418, 1_906_079, 2_028_323, 2_001_731,
    1_789_316, 1_927_499, 2_178_517, 2_354_724, 2_407_885,
    2_464_859, 3_221_836, 3_621_049, 3_099_588, 3_313_893
]
gastos_reales = [
    1_273_436, 1_528_583, 1_710_666, 1_842_753, 2_071_892,
    2_218_377, 2_469_359, 2_823_443, 3_140_716, 3_425_533,
    3_597_737, 4_102_569, 4_761_041, 5_318_713, 5_873_795
]

# Crear DataFrame histórico
df_hist = pd.DataFrame({
    'ingreso': ingresos_reales,
    'gasto': gastos_reales
}, index=años_hist)
df_hist['resultado'] = df_hist['ingreso'] - df_hist['gasto']

# ==============================
# PROYECCIÓN LINEAL PARA 2026-2030
# ==============================
X_hist = np.array(años_hist).reshape(-1, 1)
reg_ing = LinearRegression().fit(X_hist, df_hist['ingreso'].values)
reg_gas = LinearRegression().fit(X_hist, df_hist['gasto'].values)

X_proy = np.array(años_proy).reshape(-1, 1)
ing_proy = reg_ing.predict(X_proy)
gas_proy = reg_gas.predict(X_proy)

# Crear DataFrame de proyección (sin reforma)
df_proy = pd.DataFrame({
    'ingreso': ing_proy,
    'gasto': gas_proy
}, index=años_proy)
df_proy['resultado'] = df_proy['ingreso'] - df_proy['gasto']

# Concatenar todo
df_total = pd.concat([df_hist, df_proy])
print("Proyección lineal 2026-2030 (sin reforma):")
print(df_total.loc[2026:2030, ['ingreso','gasto','resultado']])

# ==============================
# PATRIMONIO BASE 2011 (reconstrucción con interés 9% desde 2011)
# ==============================
# Patrimonio 2023 conocido: USD 741 MM -> Gs
pat_2023_gs = 741 * tipo_cambio  # 5.409.300 millones Gs

# Primero reconstruimos el patrimonio sin interés para tener el valor de 2011 (igual que antes)
pat_sin_interes = {}
pat_sin_interes[2023] = pat_2023_gs
for año in range(2022, 2010, -1):
    pat_sin_interes[año] = pat_sin_interes[año+1] - df_hist.loc[año+1, 'resultado']
for año in range(2024, 2031):
    pat_sin_interes[año] = pat_sin_interes[año-1] + df_total.loc[año, 'resultado']

pat_2011_sin_interes = pat_sin_interes[2011]

# Ahora calculamos el patrimonio con interés acumulado desde 2011 hasta 2025 (usando resultados reales)
pat_con_interes = {}
pat_con_interes[2011] = pat_2011_sin_interes
for año in range(2012, 2026):
    pat_con_interes[año] = pat_con_interes[año-1] * (1 + tasa_interes) + df_hist.loc[año, 'resultado']

# A partir de 2026, usaremos los resultados proyectados (sin reforma) para el escenario base
for año in años_proy:
    pat_con_interes[año] = pat_con_interes[año-1] * (1 + tasa_interes) + df_proy.loc[año, 'resultado']

print(f"\nPatrimonio 2011 (con interés acumulado hasta 2025): {pat_con_interes[2011]/tipo_cambio:,.0f} USD MM")

# ==============================
# ESCENARIOS DE REFORMA DESDE 2026
# ==============================
# Factores de ajuste respecto a tasa base 16%
factores = {tasa: tasa / tasa_base for tasa in tasas_reforma}

# Almacenar resultados
resultados_reforma = {}

for tasa in tasas_reforma:
    factor = factores[tasa]
    # Ingresos ajustados desde 2026 (mantenemos históricos sin cambio)
    ingresos_ajustados = df_total['ingreso'].copy()
    ingresos_ajustados.loc[años_proy] = df_proy['ingreso'] * factor
    
    # Resultados ajustados
    resultados_ajustados = ingresos_ajustados - df_total['gasto']
    
    # Reconstruir patrimonio con interés desde 2011, pero usando resultados ajustados solo desde 2026
    # Para 2011-2025 usamos los mismos que en el escenario base (con interés)
    pat_reforma = {}
    pat_reforma[2011] = pat_con_interes[2011]  # mismo punto de partida
    
    for año in range(2012, 2026):
        # Hasta 2025, los resultados son los reales (sin cambio de tasa)
        pat_reforma[año] = pat_reforma[año-1] * (1 + tasa_interes) + df_hist.loc[año, 'resultado']
    
    for año in años_proy:
        pat_reforma[año] = pat_reforma[año-1] * (1 + tasa_interes) + resultados_ajustados.loc[año]
    
    resultados_reforma[tasa] = {
        'patrimonio': pat_reforma,
        'ingresos': ingresos_ajustados,
        'resultados': resultados_ajustados
    }

# También guardamos el escenario base (sin reforma) con interés para comparar
resultados_reforma['base'] = {
    'patrimonio': pat_con_interes,
    'ingresos': df_total['ingreso'],
    'resultados': df_total['resultado']
}

# ==============================
# GRÁFICOS COMPARATIVOS
# ==============================
plt.style.use('seaborn-v0_8-darkgrid')
fig, axes = plt.subplots(2, 2, figsize=(16, 12))
axes = axes.flatten()

colores = {'base': 'blue', 0.19: 'green', 0.22: 'orange', 0.25: 'red'}
etiquetas = {'base': 'Base 16% (sin reforma)', 0.19: 'Reforma 19% desde 2026', 
             0.22: 'Reforma 22% desde 2026', 0.25: 'Reforma 25% desde 2026'}

# Gráfico 1: Evolución del patrimonio (USD MM)
ax = axes[0]
for key in ['base', 0.19, 0.22, 0.25]:
    pat = resultados_reforma[key]['patrimonio']
    años = list(pat.keys())
    valores_usd = [pat[a] / tipo_cambio for a in años]
    ax.plot(años, valores_usd, 'o-', linewidth=2, markersize=3, 
            color=colores[key], label=etiquetas[key])
ax.axhline(y=0, color='black', linestyle='--', linewidth=0.5)
ax.axvline(x=2025.5, color='gray', linestyle=':', linewidth=1, label='Inicio reforma')
ax.set_title('Patrimonio total con interés 9% anual', fontsize=14, fontweight='bold')
ax.set_xlabel('Año')
ax.set_ylabel('Millones USD')
ax.legend(loc='upper left', fontsize=9)
ax.grid(True, alpha=0.3)
ax.set_xlim(2010, 2031)

# Gráfico 2: Patrimonio en 2030
ax = axes[1]
x = np.arange(len(tasas_reforma) + 1)
etiquetas_x = ['Base'] + [f'{int(t*100)}%' for t in tasas_reforma]
valores_2030 = [resultados_reforma['base']['patrimonio'][2030]/tipo_cambio] + \
               [resultados_reforma[t]['patrimonio'][2030]/tipo_cambio for t in tasas_reforma]
colores_bar = ['blue', 'green', 'orange', 'red']
bars = ax.bar(x, valores_2030, color=colores_bar, edgecolor='black')
ax.set_xticks(x)
ax.set_xticklabels(etiquetas_x)
ax.set_title('Patrimonio en 2030 (millones USD)', fontsize=14, fontweight='bold')
ax.set_ylabel('Millones USD')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, valores_2030):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 50, f'{val:,.0f}', 
            ha='center', fontweight='bold', fontsize=10)

# Gráfico 3: Resultado anual en 2030 (miles de millones Gs)
ax = axes[2]
resultados_2030 = [resultados_reforma['base']['resultados'].loc[2030]/1000] + \
                  [resultados_reforma[t]['resultados'].loc[2030]/1000 for t in tasas_reforma]
bars = ax.bar(x, resultados_2030, color=colores_bar, edgecolor='black')
ax.axhline(y=0, color='black', linestyle='--')
ax.set_xticks(x)
ax.set_xticklabels(etiquetas_x)
ax.set_title('Resultado anual 2030 (miles de millones Gs)', fontsize=14, fontweight='bold')
ax.set_ylabel('Miles de MM Gs')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, resultados_2030):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + (20 if val>0 else -30), 
            f'{val:,.0f}', ha='center', fontweight='bold', fontsize=10)

# Gráfico 4: Diferencia de patrimonio respecto al base en 2030
ax = axes[3]
diferencias = [0] + [resultados_reforma[t]['patrimonio'][2030]/tipo_cambio - resultados_reforma['base']['patrimonio'][2030]/tipo_cambio 
                     for t in tasas_reforma]
bars = ax.bar(x, diferencias, color=colores_bar, edgecolor='black')
ax.axhline(y=0, color='black', linestyle='-')
ax.set_xticks(x)
ax.set_xticklabels(etiquetas_x)
ax.set_title('Ganancia patrimonial respecto al base en 2030 (USD MM)', fontsize=14, fontweight='bold')
ax.set_ylabel('Millones USD')
ax.grid(True, alpha=0.3, axis='y')
for bar, val in zip(bars, diferencias):
    ax.text(bar.get_x() + bar.get_width()/2, bar.get_height() + 10 if val>0 else -20, 
            f'{val:+,.0f}', ha='center', fontweight='bold', fontsize=10)

plt.suptitle('Impacto de reforma con aumento de tasa de cotización desde 2026 (interés 9% anual)', 
             fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# ==============================
# TABLA COMPARATIVA (millones USD)
# ==============================
print("\n" + "="*90)
print("PATRIMONIO EN MILLONES USD - COMPARATIVA REFORMA DESDE 2026")
print("="*90)
print(f"{'Año':>6} | {'Base 16%':>15} | {'19% desde 2026':>15} | {'22% desde 2026':>15} | {'25% desde 2026':>15}")
print("-"*80)
for año in range(2020, 2031):
    val_base = resultados_reforma['base']['patrimonio'][año]/tipo_cambio
    val_19 = resultados_reforma[0.19]['patrimonio'][año]/tipo_cambio
    val_22 = resultados_reforma[0.22]['patrimonio'][año]/tipo_cambio
    val_25 = resultados_reforma[0.25]['patrimonio'][año]/tipo_cambio
    print(f"{año:6d} | {val_base:15,.0f} | {val_19:15,.0f} | {val_22:15,.0f} | {val_25:15,.0f}")

print("\n" + "="*90)
print("DÉFICIT ANUAL EN 2030 (miles de millones Gs)")
print("="*90)
for tasa in [0.16] + tasas_reforma:
    if tasa == 0.16:
        deficit = resultados_reforma['base']['resultados'].loc[2030]/1000
        print(f"Base 16% (sin reforma): {deficit:,.0f} mil MM Gs")
    else:
        deficit = resultados_reforma[tasa]['resultados'].loc[2030]/1000
        print(f"Reforma {int(tasa*100)}% desde 2026: {deficit:,.0f} mil MM Gs")
```
